```python
# Crear una serie de tres colores diferentes y visualizarla
colores = pd.Series(['rojo', 'amarillo','verde'])
colores
```

```python
# Crear una serie de tipos de autos
lista_autos = pd.Series(['sedan','SUV','Pick up'])
lista_autos
```

```python
# Combinar las series en un dataframe
tabla_autos = pd.DataFrame({'Tipo de auto':lista_autos, 'Color':colores})
tabla_autos
```

```python
# Crear tabla cruzada
pd.crosstab(ventas_autos['Fabricante'], ventas_autos['Puertas'])
```

```python
# Agrupar columnas por fabricante y buscando el valor medio de los columnas numericas
ventas_autos.groupby(["Fabricante"]).mean(numeric_only=True)
```
